```python
# This code generates figure S1-3 from supplementary material. It need the data
# from the spacings related to the oscillations as calculated from the
# reflectance spectra GC-spacing.csv and DS-spacing.csv
import numpy as np
import matplotlib.pyplot as plt

dataGC = np.loadtxt("GC-spacing.csv")
dataDS = np.loadtxt("DS-spacing.csv")

plt.boxplot([dataGC,dataDS],labels=["G.costaricensis","D. solitarius"])
plt.title("Distances Related to Oscillations")
plt.ylabel("Distance ($\mu                        $ m)")
plt.xlabel("Species")
plt.savefig('distances.tiff', dpi=300)
plt.show()
```
